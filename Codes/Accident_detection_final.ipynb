```python
from ultralytics import YOLO

model = YOLO('yolov8s.pt')

model.train(
    data="/content/accident_dataset/accident_data.yaml",
    epochs=50,
    batch=16,
    patience=20,
    device='0'
)
```

```python
#loading the best trained model for future predictions
model = YOLO('/content/runs/detect/train2/weights/best.pt')

img = '/content/acccident.jpg'

pred = model.predict(img,save=True)
```
