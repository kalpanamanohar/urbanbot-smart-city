```python
yaml_content = """train: /content/potholes_dataset/data/images
val: /content/potholes_dataset/data/images
nc: 1
names: ['pothole']"""

with open('/content/potholes_dataset/data/potholes.yaml', 'w') as file:
    file.write(yaml_content)
```

```python
!pip install ultralytics -q
```

```python
from ultralytics import YOLO

model = YOLO('yolov8s.pt')   # Upgrade model

model.train(
    data="/content/potholes_dataset/data/potholes.yaml",
    epochs=200,
    imgsz=960,
    batch=16,
    patience=50,
    device=0,
    workers=4,
    optimizer='AdamW'
)
```

```python
#loading the best trained model for future predictions
model = YOLO('/content/runs/detect/train/weights/best.pt')

img = 'potholes.jpg'

pred = model.predict(img,save= True)
```
